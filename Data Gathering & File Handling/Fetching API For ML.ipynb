```python
import pandas as pd
import requests
```

```python
response = requests.get('https://api.themoviedb.org/3/movie/popular?api_key=7c5774dfa4a21b5e2c31126ada7daa1c&language=en-US&page=1')

df_products = pd.DataFrame(response.json()['results']) [['id', 'title', 'popularity', 'release_date', 'vote_average', 'vote_count']]
df_products.head(2)
```

```python
all_pages_data = []


```

```python
for i in range(1, 429):
    response = requests.get('https://api.themoviedb.org/3/movie/popular?api_key=7c5774dfa4a21b5e2c31126ada7daa1c&language=en-US&page={}'.format(i))
    temp = pd.DataFrame(response.json()['results'])[['id', 'title', 'popularity', 'release_date', 'vote_average', 'vote_count']]
    all_pages_data.append(temp)


df = pd.concat(all_pages_data, ignore_index=True)
df
```
