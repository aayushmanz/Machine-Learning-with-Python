# Scraping : 

```python
product = soup.find_all('div', class_='product-wrapper card-body')
len(product)
```

```python
title = []
price = []
desc = []
review = []

for i in product:
    title.append(i.find('a', class_='title').text.strip().split('...')[0])
    price.append(i.find('span', itemprop='price').text.strip())
    desc.append(i.find('p', itemprop='description').text.strip())
    review.append(i.find('span', itemprop='reviewCount').text.strip())
```

```python
d = {
    'title':title,
    'price':price,
    'review' :review,
    'description':desc
}

df = pd.DataFrame(d)
```
