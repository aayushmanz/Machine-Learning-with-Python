```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
webpage = requests.get('https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops',headers=headers).text
```

```python

soup = BeautifulSoup(webpage, 'html.parser')
```

```python
print(soup.title)
```

```python
soup.find_all('h4')[1].text.strip()
```

```python
for i in soup.find_all('a', class_='title'):
    print(i.text.strip().split('...')[0])
```

```python
soup.find_all('span', itemprop="price")[1].text
```

```python
for i in soup.find_all('span', itemprop="price"):
    print(i.text)
```
