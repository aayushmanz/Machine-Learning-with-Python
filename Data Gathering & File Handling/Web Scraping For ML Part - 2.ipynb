```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python

for i in range(1,17):
  data = requests.get('https://webscraper.io/test-sites/pagination?page={}'.format(i))

```

```python
soup = BeautifulSoup(data.text, 'html.parser')
    
```

```python
for i in soup.find_all('h3', itemprop='name'):
    print(i.text.strip())
```
