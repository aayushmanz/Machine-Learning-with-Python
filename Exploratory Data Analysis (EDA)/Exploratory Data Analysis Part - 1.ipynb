```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/workspaces/Machine-Learning-with-Python/Exploratory Data Analysis (EDA)/train.csv')
```

```python
df.head()
```

## 1. Categorical Data

### a. Countplot

```python
sns.countplot(data=df, x='Survived', hue='Survived')


```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x=df['Pclass'], hue=df['Pclass'])
```

```python
sns.countplot(x=df['Embarked'], hue=df['Embarked'])
```

### b. piechart 

```python
df['Survived'].value_counts().plot(kind='pie', autopct="%1.1f%%")
```

```python
df['Pclass'].value_counts().plot(kind='pie', autopct="%1.1f%%")
```

```python
df['Sex'].value_counts().plot(kind='pie', autopct="%1.1f%%")
```

## 2. Numerical Data 

### a. Histogram

```python
sns.histplot(x=df['Age'], hue=df['Survived'], bins=50)
```

### b. Distplot

```python
sns.distplot(x=df['Age'])
plt.show()
```

### c. Boxplot

```python
sns.boxplot(data=df, x='Fare')
```

```python
sns.boxplot(data=df, x='Age')
```
